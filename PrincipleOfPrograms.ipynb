```python
"""
Author: Ryon Connery
Date: September 1, 2026

Program: Simple Purchase Calculation Program

Description:
This program prompts the user to enter the unit price of an item and the
quantity purchased. It calculates the original purchase price, applies
either a 5 percent or 3 percent discount based on the purchase amount,
calculates a 6 percent tax and 2 percent shipping cost, and displays the
final amount paid.

Variables:
unit_price       - Price of one item entered by the user.
quantity         - Number of items purchased.
original_price   - Unit price multiplied by quantity.
discount_rate    - Discount percentage selected by the program.
discount_amount  - Dollar amount of the discount.
total_after_discount - Purchase total after the discount is subtracted.
tax_amount       - Six percent tax on the discounted total.
shipping_cost    - Two percent shipping cost on the discounted total.
total_amount_paid - Final amount including tax and shipping.

Program Flow:
1. Display a welcome message.
2. Request the unit price and quantity.
3. Calculate the original purchase price.
4. Select the correct discount rate using an if/else statement.
5. Calculate the discount and discounted total.
6. Calculate tax and shipping.
7. Calculate and display the final amount paid.
"""

# Constants
LOW_DISCOUNT_RATE = 0.03
HIGH_DISCOUNT_RATE = 0.05
TAX_RATE = 0.06
SHIPPING_RATE = 0.02

# Welcome message
print("Welcome to the Simple Purchase Calculation Program")
print("Programmed by Ryon Connery")
print()

# User input
unit_price = float(input("Enter the price per item: $"))
quantity = int(input("Enter the quantity purchased: "))

# Calculate original price
original_price = unit_price * quantity

# Determine the discount rate
if original_price >= 100:
    discount_rate = HIGH_DISCOUNT_RATE
else:
    discount_rate = LOW_DISCOUNT_RATE

# Calculate discount
discount_amount = original_price * discount_rate
total_after_discount = original_price - discount_amount

# Calculate tax and shipping
tax_amount = total_after_discount * TAX_RATE
shipping_cost = total_after_discount * SHIPPING_RATE

# Calculate final amount paid
total_amount_paid = total_after_discount + tax_amount + shipping_cost

# Display results
print()
print("Purchase Summary")
print("----------------------------")
print("Original Price: $", original_price)
print("Discount Rate:", discount_rate * 100, "%")
print("Discount Amount: $", discount_amount)
print("Total After Discount: $", total_after_discount)
print("Tax Amount: $", tax_amount)
print("Shipping Cost: $", shipping_cost)
print("Total Amount Paid: $", total_amount_paid)
print()
print("Thank you for using the program, Ryon Connery.")
```
